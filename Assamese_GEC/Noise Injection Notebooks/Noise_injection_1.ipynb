```python
import pandas as pd
import random
import re

TARGET_TENSE = 1173
random.seed(42)

remaining_df = pd.read_csv("/kaggle/working/remaining_candidates.csv")

def inject_tense_error(sentence, tense_map):
    words = sentence.split()

    for i, word in enumerate(words):
        clean_word = re.sub(r"[।,!?;:]", "", word)

        if clean_word in tense_map:
            replacement = random.choice(tense_map[clean_word])
            words[i] = word.replace(clean_word, replacement)
            return " ".join(words), True

    return sentence, False

remaining_shuffled = remaining_df.sample(frac=1, random_state=43).reset_index(drop=True)

tense_rows = []

for idx, row in remaining_shuffled.iterrows():
    if len(tense_rows) == TARGET_TENSE:
        break

    original_sentence = str(row["sentence"])
    noisy_sentence, success = inject_tense_error(original_sentence, tense_map)

    if success:
        tense_rows.append({
            
            "sentence": noisy_sentence,
            "error_label": 3
        })

final_tense_df = pd.DataFrame(tense_rows)

final_tense_df.to_csv(
    "/kaggle/working/tense_error_sentences.csv",
    index=False,
    encoding="utf-8"
)

print("Final tense errors:", len(final_tense_df))
print("Still missing:", TARGET_TENSE - len(final_tense_df))
print(final_tense_df.head())
```

```python
import pandas as pd
import random
import re

random.seed(42)

TARGET_TENSE = 1173

# Load full 51k dataset
df = pd.read_csv("/kaggle/working/51k_noise_candidates.csv")

# Add label column to full 51k
df["error_label"] = 0

# Load tense table
tense_table = pd.read_excel(
    "/kaggle/input/datasets/shruteesaha/assamese-tense-forms/Assamese_Tense_Person_Extraction.xlsx"
)

# Build replacement map
tense_map = {}

for _, row in tense_table.iterrows():
    forms = [
        str(row["Present"]).strip(),
        str(row["Past"]).strip(),
        str(row["Future"]).strip()
    ]
    forms = [f for f in forms if f and f.lower() != "nan"]

    for form in forms:
        tense_map[form] = [x for x in forms if x != form]


def inject_tense_error(sentence, tense_map):
    words = sentence.split()

    for i, word in enumerate(words):
        clean_word = re.sub(r"[।,!?;:]", "", word)

        if clean_word in tense_map:
            replacement = random.choice(tense_map[clean_word])
            words[i] = word.replace(clean_word, replacement)
            return " ".join(words), True

    return sentence, False


# Randomly shuffle 51k rows
shuffled_indices = df.sample(frac=1, random_state=42).index

success_count = 0

for idx in shuffled_indices:
    if success_count == TARGET_TENSE:
        break

    original_sentence = str(df.loc[idx, "sentence"])
    noisy_sentence, success = inject_tense_error(original_sentence, tense_map)

    if success:
        df.loc[idx, "sentence"] = noisy_sentence
        df.loc[idx, "error_label"] = 3
        success_count += 1

# Save FULL 51k dataset, not separate 1173 file
df.to_csv(
    "/kaggle/working/51k_after_tense_injection.csv",
    index=False,
    encoding="utf-8"
)

print("Total rows:", len(df))
print("Tense errors injected:", success_count)
print("Label counts:")
print(df["error_label"].value_counts())
```

```python
import pandas as pd
import random
import re

random.seed(42)

TARGET_PRONOUN = round((4 / 343) * 51000)
print("Target pronoun errors:", TARGET_PRONOUN)

# Load full 51k after tense injection
df = pd.read_csv("/kaggle/working/51k_after_tense_injection.csv")

# Load pronoun rules
pronoun_rules = pd.read_excel(
    "/kaggle/input/datasets/shruteesaha/assamese-pronoun-rules/assamese_pronoun_rules.xlsx",
    sheet_name="Singular_Plural_Pairs"
)

# Build pronoun replacement map: singular <-> plural
pronoun_map = {}

for _, row in pronoun_rules.iterrows():
    singular = str(row["এক-বচন ৰূপ"]).strip()
    plural = str(row["বহু-বচন ৰূপ"]).strip()

    if singular and plural and singular != "nan" and plural != "nan":
        pronoun_map[singular] = plural
        pronoun_map[plural] = singular


def inject_pronoun_error(sentence, pronoun_map):
    words = sentence.split()

    for i, word in enumerate(words):
        clean_word = re.sub(r"[।,!?;:]", "", word)

        if clean_word in pronoun_map:
            replacement = pronoun_map[clean_word]
            words[i] = word.replace(clean_word, replacement)
            return " ".join(words), True

    return sentence, False


shuffled_indices = df.sample(frac=1, random_state=43).index

success_count = 0

for idx in shuffled_indices:
    if success_count == TARGET_PRONOUN:
        break

    original_sentence = str(df.loc[idx, "sentence"])
    noisy_sentence, success = inject_pronoun_error(original_sentence, pronoun_map)

    if success:
        df.loc[idx, "sentence"] = noisy_sentence

        old_label = str(df.loc[idx, "error_label"])

        if old_label == "0":
            df.loc[idx, "error_label"] = "6"
        elif "6" not in old_label.split(","):
            df.loc[idx, "error_label"] = old_label + ",6"

        success_count += 1


df.to_csv(
    "/kaggle/working/51k_after_tense_pronoun_injection.csv",
    index=False,
    encoding="utf-8"
)

print("Total rows:", len(df))
print("Pronoun errors injected:", success_count)
print("Label counts:")
print(df["error_label"].value_counts())
```

```python
import pandas as pd
import random

INPUT_PATH = "/kaggle/working/51k_after_tense_pronoun_injection.csv"
PRATYAYA_EXCEL_PATH = "/kaggle/input/datasets/shruteesaha/pratyaya-error/PRATYAY_ERROR.xlsx"
OUTPUT_PATH = "/kaggle/working/51k_after_tense_pronoun_pratyaya_injection.csv"
LOG_PATH = "/kaggle/working/pratyaya_injection_log.csv"

random.seed(42)

df = pd.read_csv(INPUT_PATH)
rules_df = pd.read_excel(PRATYAYA_EXCEL_PATH)

rules_df.columns = rules_df.columns.str.strip()
rules_df = rules_df[["Correct Word", "Incorrect Pratyaya Form"]].copy()

rules_df["Correct Word"] = rules_df["Correct Word"].astype(str).str.strip()
rules_df["Incorrect Pratyaya Form"] = rules_df["Incorrect Pratyaya Form"].astype(str).str.strip()

rules_df = rules_df[
    (rules_df["Correct Word"] != "") &
    (rules_df["Incorrect Pratyaya Form"] != "") &
    (rules_df["Correct Word"].str.lower() != "nan") &
    (rules_df["Incorrect Pratyaya Form"].str.lower() != "nan")
]

pratyaya_map = dict(zip(
    rules_df["Correct Word"],
    rules_df["Incorrect Pratyaya Form"]
))

TARGET_PRATYAYA = round(len(df) * (17 / 343))

print("Total rows:", len(df))
print("Total pratyaya rules:", len(pratyaya_map))
print("Target pratyaya errors:", TARGET_PRATYAYA)

candidate_rows = []

for idx, sent in df["sentence"].astype(str).items():
    matches = []

    for correct, incorrect in pratyaya_map.items():
        if correct in sent:
            matches.append((correct, incorrect))

    if matches:
        candidate_rows.append((idx, matches))

print("Candidate rows found:", len(candidate_rows))

selected_rows = random.sample(
    candidate_rows,
    min(TARGET_PRATYAYA, len(candidate_rows))
)

log = []
injected_count = 0

for idx, matches in selected_rows:
    old_sentence = str(df.at[idx, "sentence"])
    old_label = str(df.at[idx, "error_label"]).strip()

    correct, incorrect = random.choice(matches)
    new_sentence = old_sentence.replace(correct, incorrect, 1)

    if new_sentence != old_sentence:
        df.at[idx, "sentence"] = new_sentence

        if old_label == "0":
            new_label = "9"
        else:
            labels = [x.strip() for x in old_label.split(",") if x.strip()]
            if "9" not in labels:
                labels.append("9")
            labels = sorted(labels, key=int)
            new_label = ",".join(labels)

        df.at[idx, "error_label"] = new_label

        injected_count += 1

        log.append({
            "row_index": idx,
            "correct_word_replaced": correct,
            "incorrect_form_inserted": incorrect,
            "old_label": old_label,
            "new_label": new_label,
            "original_sentence": old_sentence,
            "modified_sentence": new_sentence
        })

print("Actually injected:", injected_count)

df.to_csv(OUTPUT_PATH, index=False)
pd.DataFrame(log).to_csv(LOG_PATH, index=False)

print("\nLabel counts:")
print(df["error_label"].value_counts())

print("\nSaved final dataset:", OUTPUT_PATH)
print("Saved log:", LOG_PATH)
```

```python
import pandas as pd
import random

INPUT_PATH = "/kaggle/input/datasets/shruteesaha/51k-after-tense-pronoun-pratyaya-injection/51k_after_tense_pronoun_pratyaya_injection.csv"
NUMBER_RULES_PATH = "/kaggle/input/datasets/shruteesaha/number-agreement-noun-based/Number_agreement_noun_based.xlsx"
OUTPUT_PATH = "/kaggle/working/51k_after_tense_pronoun_pratyaya_number_injection.csv"

NUMBER_LABEL = 5
TARGET_RATIO = 14 / 343
RANDOM_SEED = 42
random.seed(RANDOM_SEED)

df = pd.read_csv(INPUT_PATH)

rules_df = pd.read_excel(NUMBER_RULES_PATH)

rules_df = rules_df.dropna(subset=["Correct", "Incorrect"])

rules_df = rules_df[
    ~(
        rules_df["Correct"].astype(str).str.strip().isin(["লো", "লোবোৰ"]) |
        rules_df["Incorrect"].astype(str).str.strip().isin(["লো", "লোবোৰ"])
    )
]

rules = list(
    zip(
        rules_df["Correct"].astype(str).str.strip(),
        rules_df["Incorrect"].astype(str).str.strip()
    )
)

print("Total rules:", len(rules))

def inject_number_error(sentence):

    applicable = []

    for correct, incorrect in rules:
        if correct in sentence:
            applicable.append((correct, incorrect))

    if not applicable:
        return None

    correct, incorrect = random.choice(applicable)

    noisy_sentence = sentence.replace(correct, incorrect, 1)

    if noisy_sentence == sentence:
        return None

    return noisy_sentence


target_count = round(len(df) * TARGET_RATIO)

candidate_indices = list(df.index)
random.shuffle(candidate_indices)

modified = 0

for idx in candidate_indices:

    if modified >= target_count:
        break

    sentence = str(df.at[idx, "sentence"])

    noisy_sentence = inject_number_error(sentence)

    if noisy_sentence is None:
        continue

    df.at[idx, "sentence"] = noisy_sentence

    old_label = str(df.at[idx, "error_label"]).strip()


    if old_label == "0":

        df.at[idx, "error_label"] = str(NUMBER_LABEL)

    else:

        labels = [x.strip() for x in old_label.split(",")]

        if str(NUMBER_LABEL) not in labels:
            labels.append(str(NUMBER_LABEL))

        # sort numerically for consistency
        labels = sorted(labels, key=int)

        df.at[idx, "error_label"] = ",".join(labels)

    modified += 1


df = df[["sentence", "error_label"]]

df.to_csv(OUTPUT_PATH, index=False)

print(f"\nTarget number errors : {target_count}")
print(f"Injected successfully : {modified}")

print("\nLabel Distribution:")
print(df["error_label"].value_counts().head(20))

print(f"\nSaved to : {OUTPUT_PATH}")

print("\nSample:")
print(df.head())
```

```python
import pandas as pd
import random
import re
from collections import defaultdict

# =========================
# PATHS
# =========================
INPUT_PATH = "/kaggle/input/datasets/shruteesaha/51k-till-pronoun/51k_after_tense_pronoun_pratyaya_number_person_injection.csv"
VERB_RULES_PATH = "/kaggle/input/datasets/shruteesaha/verb-grammar-rule-semi-final/VERB_grammar_rule_semi_final.xlsx"

OUTPUT_PATH = "/kaggle/working/51k_after_tense_pronoun_pratyaya_number_verb_injection.csv"
REPORT_PATH = "/kaggle/working/verb_injection_report.xlsx"

# =========================
# SETTINGS
# =========================
VERB_LABEL = 2
TARGET_RATIO = 33 / 343
RANDOM_SEED = 43

random.seed(RANDOM_SEED)

# =========================
# LOAD DATA
# =========================
df = pd.read_csv(INPUT_PATH)
df["sentence"] = df["sentence"].astype(str)
df["error_label"] = df["error_label"].astype(str)

rules_df = pd.read_excel(VERB_RULES_PATH)

rules_df = rules_df.dropna(subset=["Correct", "Incorrect"])

rules_df["Correct"] = rules_df["Correct"].astype(str).str.strip()
rules_df["Incorrect"] = rules_df["Incorrect"].astype(str).str.strip()

# remove accidental header rows
rules_df = rules_df[
    (rules_df["Correct"] != "Correct") &
    (rules_df["Incorrect"] != "Incorrect")
].copy()

# remove empty / too short risky rules
rules_df = rules_df[
    (rules_df["Correct"].str.len() >= 3) &
    (rules_df["Incorrect"].str.len() >= 2)
].copy()

# remove duplicate rules
rules_df = rules_df.drop_duplicates(subset=["Correct", "Incorrect"])

rules = list(zip(rules_df["Correct"], rules_df["Incorrect"]))

print("Total verb rules after cleaning:", len(rules))

# =========================
# SAFE PHRASE MATCHING
# =========================
def make_pattern(phrase):
    """
    Matches phrase only when it is not inside another continuous token.
    Prevents:
    কৰি -> কৰিবলৈ inside 'কৰিবলৈ'
    জানি -> জানিবলৈ inside 'জানিছিল'
    """
    return rf'(?<!\S){re.escape(phrase)}(?!\S)'


def has_phrase(sentence, phrase):
    pattern = make_pattern(phrase)
    return re.search(pattern, sentence) is not None


def replace_phrase_once(sentence, correct, incorrect):
    pattern = make_pattern(correct)
    return re.sub(pattern, incorrect, sentence, count=1)


def add_label(old_label, new_label):
    old_label = str(old_label).strip()

    if old_label == "0":
        return str(new_label)

    labels = [x.strip() for x in old_label.split(",")]

    if str(new_label) not in labels:
        labels.append(str(new_label))

    labels = sorted(labels, key=int)

    return ",".join(labels)

# =========================
# BUILD CANDIDATE POOLS
# =========================
rule_candidates = {}

sentences = df["sentence"].astype(str).tolist()

for correct, incorrect in rules:

    indices = []

    for idx, sent in enumerate(sentences):

        if has_phrase(sent, correct):
            indices.append(idx)

    if indices:
        rule_candidates[(correct, incorrect)] = indices

print("Rules with matches:", len(rule_candidates))

# =========================
# TARGET COUNT
# =========================
target_count = round(len(df) * TARGET_RATIO)

print("Target verb errors:", target_count)

# =========================
# FAIR RULE-BALANCED INJECTION
# =========================
used_indices = set()
modified = 0

rule_usage = defaultdict(int)

num_rules = len(rule_candidates)

base_quota = max(1, target_count // num_rules)

print("Base quota per matched rule:", base_quota)

rule_list = list(rule_candidates.items())
random.shuffle(rule_list)

# =========================
# FIRST PASS: FAIR QUOTA PER RULE
# =========================
for (correct, incorrect), candidates in rule_list:

    if modified >= target_count:
        break

    available = [
        idx for idx in candidates
        if idx not in used_indices
    ]

    random.shuffle(available)

    quota = min(
        base_quota,
        len(available),
        target_count - modified
    )

    for idx in available[:quota]:

        sentence = str(df.at[idx, "sentence"])

        noisy_sentence = replace_phrase_once(
            sentence,
            correct,
            incorrect
        )

        if noisy_sentence == sentence:
            continue

        df.at[idx, "sentence"] = noisy_sentence

        df.at[idx, "error_label"] = add_label(
            df.at[idx, "error_label"],
            VERB_LABEL
        )

        used_indices.add(idx)
        rule_usage[(correct, incorrect)] += 1
        modified += 1

# =========================
# SECOND PASS: FILL REMAINING
# =========================
if modified < target_count:

    all_remaining_candidates = []

    for (correct, incorrect), candidates in rule_candidates.items():

        for idx in candidates:

            if idx not in used_indices:

                all_remaining_candidates.append(
                    (idx, correct, incorrect)
                )

    random.shuffle(all_remaining_candidates)

    for idx, correct, incorrect in all_remaining_candidates:

        if modified >= target_count:
            break

        sentence = str(df.at[idx, "sentence"])

        noisy_sentence = replace_phrase_once(
            sentence,
            correct,
            incorrect
        )

        if noisy_sentence == sentence:
            continue

        df.at[idx, "sentence"] = noisy_sentence

        df.at[idx, "error_label"] = add_label(
            df.at[idx, "error_label"],
            VERB_LABEL
        )

        used_indices.add(idx)
        rule_usage[(correct, incorrect)] += 1
        modified += 1

# =========================
# SAVE DATASET
# =========================
df = df[["sentence", "error_label"]]

df.to_csv(OUTPUT_PATH, index=False)

print("\nTarget verb errors:", target_count)
print("Injected successfully:", modified)
print("Saved to:", OUTPUT_PATH)

print("\nLabel Distribution:")
print(df["error_label"].value_counts().head(40))

# =========================
# SAVE REPORT
# =========================
report_rows = []

for correct, incorrect in rules:

    candidate_count = len(
        rule_candidates.get((correct, incorrect), [])
    )

    injected_count = rule_usage.get(
        (correct, incorrect),
        0
    )

    report_rows.append({
        "Correct": correct,
        "Incorrect": incorrect,
        "Candidate_Freq": candidate_count,
        "Injected_Count": injected_count
    })

report_df = pd.DataFrame(report_rows)

report_df = report_df.sort_values(
    "Injected_Count",
    ascending=False
)

summary_df = pd.DataFrame({
    "Metric": [
        "Dataset Size",
        "Target Verb Errors",
        "Injected Verb Errors",
        "Total Rules",
        "Rules With Matches",
        "Rules Used For Injection"
    ],
    "Value": [
        len(df),
        target_count,
        modified,
        len(rules),
        len(rule_candidates),
        (report_df["Injected_Count"] > 0).sum()
    ]
})

with pd.ExcelWriter(REPORT_PATH) as writer:
    summary_df.to_excel(
        writer,
        sheet_name="Summary",
        index=False
    )

    report_df.to_excel(
        writer,
        sheet_name="Rule_Usage",
        index=False
    )

print("Report saved to:", REPORT_PATH)

print("\nTop used rules:")
print(report_df.head(30).to_string(index=False))

# =========================
# QUICK CHECK FOR BAD DOUBLE SUFFIXES
# =========================
bad_patterns = [
    "কৰিবলৈবলৈ",
    "জানিবলৈছিল",
    "পঢ়িবলৈবলৈ",
    "লিখিবলৈবলৈ",
    "খাবলৈবলৈ",
    "যাবলৈবলৈ",
    "আহিবলৈবলৈ"
]

print("\nBad pattern check:")

for pat in bad_patterns:

    count = df["sentence"].astype(str).str.contains(
        pat,
        regex=False,
        na=False
    ).sum()

    print(pat, ":", count)
```

```python
import pandas as pd
import random
import re
from collections import defaultdict

# =========================
# PATHS
# =========================
INPUT_PATH = "/kaggle/input/datasets/shruteesaha/51k-till-verb/51k_after_tense_pronoun_pratyaya_number_verb_injection.csv"

OUTPUT_PATH = "/kaggle/working/51k_after_punctuation_injection.csv"
REPORT_PATH = "/kaggle/working/punctuation_injection_report.xlsx"

# =========================
# SETTINGS
# =========================
PUNCT_LABEL = 11
TARGET_RATIO = 16 / 343
RANDOM_SEED = 44

random.seed(RANDOM_SEED)

# =========================
# LOAD DATA
# =========================
df = pd.read_csv(INPUT_PATH)
df["sentence"] = df["sentence"].astype(str)
df["error_label"] = df["error_label"].astype(str)

# =========================
# LABEL HANDLING
# =========================
def add_label(old_label, new_label):
    old_label = str(old_label).strip()

    if old_label == "0":
        return str(new_label)

    labels = [x.strip() for x in old_label.split(",") if x.strip()]

    if str(new_label) not in labels:
        labels.append(str(new_label))

    labels = sorted(labels, key=int)
    return ",".join(labels)

# =========================
# PUNCTUATION ERROR FUNCTIONS
# =========================
def remove_final_punctuation(s):
    return re.sub(r"[।?!]+$", "", s.strip())

def replace_final_danda_with_comma(s):
    return re.sub(r"।$", ",", s.strip())

def replace_question_with_danda(s):
    return re.sub(r"\?$", "।", s.strip())

def replace_exclamation_with_danda(s):
    return re.sub(r"!$", "।", s.strip())

def duplicate_final_punctuation(s):
    s = s.strip()
    if s.endswith("।"):
        return s + "।"
    if s.endswith("?"):
        return s + "?"
    if s.endswith("!"):
        return s + "!"
    return s

def remove_comma(s):
    return s.replace(",", "", 1)

def replace_comma_with_danda(s):
    return s.replace(",", "।", 1)

def remove_semicolon(s):
    return s.replace(";", "", 1)

def replace_semicolon_with_comma(s):
    return s.replace(";", ",", 1)

def remove_colon(s):
    return s.replace(":", "", 1)

def replace_colon_with_comma(s):
    return s.replace(":", ",", 1)

def remove_hyphen(s):
    return s.replace("-", "", 1)

def replace_hyphen_with_comma(s):
    return s.replace("-", ",", 1)

def remove_dash(s):
    return s.replace("—", "", 1)

def replace_dash_with_comma(s):
    return s.replace("—", ",", 1)

def remove_quotes(s):
    for q in ['"', "“", "”", "'", "‘", "’"]:
        if q in s:
            return s.replace(q, "", 1)
    return s

def remove_bracket(s):
    for b in ["(", ")", "[", "]", "{", "}"]:
        if b in s:
            return s.replace(b, "", 1)
    return s

def replace_ellipsis_with_danda(s):
    if "......" in s:
        return s.replace("......", "।", 1)
    if "..." in s:
        return s.replace("...", "।", 1)
    return s

def remove_ellipsis(s):
    if "......" in s:
        return s.replace("......", "", 1)
    if "..." in s:
        return s.replace("...", "", 1)
    return s

def remove_asterisk(s):
    return s.replace("*", "", 1)

def add_unnecessary_comma(s):
    words = s.split()
    if len(words) < 6:
        return s

    pos = random.randint(2, len(words) - 3)

    if words[pos].endswith((",", "।", "?", "!", ";", ":", "-", "—")):
        return s

    words[pos] = words[pos] + ","
    return " ".join(words)

def add_unnecessary_danda_mid_sentence(s):
    words = s.split()
    if len(words) < 7:
        return s

    pos = random.randint(2, len(words) - 3)

    if words[pos].endswith(("।", "?", "!", ",", ";", ":", "-", "—")):
        return s

    words[pos] = words[pos] + "।"
    return " ".join(words)

# =========================
# HARD-CODED PUNCTUATION RULES
# Based on grammar book punctuation marks:
# । , ; : ! ? - — " " ( ) ...... *
# =========================
punctuation_rules = [
    ("Remove final punctuation", lambda s: bool(re.search(r"[।?!]+$", s.strip())), remove_final_punctuation),
    ("Replace final danda with comma", lambda s: s.strip().endswith("।"), replace_final_danda_with_comma),
    ("Replace question mark with danda", lambda s: s.strip().endswith("?"), replace_question_with_danda),
    ("Replace exclamation mark with danda", lambda s: s.strip().endswith("!"), replace_exclamation_with_danda),
    ("Duplicate final punctuation", lambda s: bool(re.search(r"[।?!]+$", s.strip())), duplicate_final_punctuation),

    ("Remove comma", lambda s: "," in s, remove_comma),
    ("Replace comma with danda", lambda s: "," in s, replace_comma_with_danda),

    ("Remove semicolon", lambda s: ";" in s, remove_semicolon),
    ("Replace semicolon with comma", lambda s: ";" in s, replace_semicolon_with_comma),

    ("Remove colon", lambda s: ":" in s, remove_colon),
    ("Replace colon with comma", lambda s: ":" in s, replace_colon_with_comma),

    ("Remove hyphen", lambda s: "-" in s, remove_hyphen),
    ("Replace hyphen with comma", lambda s: "-" in s, replace_hyphen_with_comma),

    ("Remove dash", lambda s: "—" in s, remove_dash),
    ("Replace dash with comma", lambda s: "—" in s, replace_dash_with_comma),

    ("Remove quotation mark", lambda s: any(q in s for q in ['"', "“", "”", "'", "‘", "’"]), remove_quotes),
    ("Remove bracket", lambda s: any(b in s for b in ["(", ")", "[", "]", "{", "}"]), remove_bracket),

    ("Replace ellipsis with danda", lambda s: "..." in s or "......" in s, replace_ellipsis_with_danda),
    ("Remove ellipsis", lambda s: "..." in s or "......" in s, remove_ellipsis),

    ("Remove asterisk", lambda s: "*" in s, remove_asterisk),

    ("Add unnecessary comma", lambda s: len(s.split()) >= 6, add_unnecessary_comma),
    ("Add unnecessary danda mid-sentence", lambda s: len(s.split()) >= 7, add_unnecessary_danda_mid_sentence),
]

# =========================
# TARGET COUNT
# =========================
target_count = round(len(df) * TARGET_RATIO)
print("Target punctuation errors:", target_count)

# =========================
# BUILD CANDIDATE POOLS
# =========================
rule_candidates = defaultdict(list)

for idx, sentence in enumerate(df["sentence"].astype(str).tolist()):

    for rule_name, condition, func in punctuation_rules:

        if condition(sentence):
            noisy = func(sentence)

            if noisy != sentence:
                rule_candidates[rule_name].append(idx)

print("Rules with matches:", len(rule_candidates))

# =========================
# FAIR RULE-BALANCED INJECTION
# =========================
used_indices = set()
modified = 0
rule_usage = defaultdict(int)

matched_rules = [
    (rule_name, condition, func)
    for rule_name, condition, func in punctuation_rules
    if rule_name in rule_candidates
]

base_quota = max(1, target_count // len(matched_rules))
print("Base quota per matched rule:", base_quota)

random.shuffle(matched_rules)

# =========================
# FIRST PASS
# =========================
for rule_name, condition, func in matched_rules:

    if modified >= target_count:
        break

    available = [
        idx for idx in rule_candidates[rule_name]
        if idx not in used_indices
    ]

    random.shuffle(available)

    quota = min(base_quota, len(available), target_count - modified)

    for idx in available[:quota]:

        sentence = str(df.at[idx, "sentence"])
        noisy_sentence = func(sentence)

        if noisy_sentence == sentence:
            continue

        df.at[idx, "sentence"] = noisy_sentence
        df.at[idx, "error_label"] = add_label(df.at[idx, "error_label"], PUNCT_LABEL)

        used_indices.add(idx)
        rule_usage[rule_name] += 1
        modified += 1

# =========================
# SECOND PASS
# =========================
if modified < target_count:

    all_remaining_candidates = []

    for rule_name, condition, func in matched_rules:
        for idx in rule_candidates[rule_name]:
            if idx not in used_indices:
                all_remaining_candidates.append((idx, rule_name, func))

    random.shuffle(all_remaining_candidates)

    for idx, rule_name, func in all_remaining_candidates:

        if modified >= target_count:
            break

        sentence = str(df.at[idx, "sentence"])
        noisy_sentence = func(sentence)

        if noisy_sentence == sentence:
            continue

        df.at[idx, "sentence"] = noisy_sentence
        df.at[idx, "error_label"] = add_label(df.at[idx, "error_label"], PUNCT_LABEL)

        used_indices.add(idx)
        rule_usage[rule_name] += 1
        modified += 1

# =========================
# SAVE DATASET
# =========================
df = df[["sentence", "error_label"]]
df.to_csv(OUTPUT_PATH, index=False)

print("\nTarget punctuation errors:", target_count)
print("Injected successfully:", modified)
print("Saved to:", OUTPUT_PATH)

print("\nLabel Distribution:")
print(df["error_label"].value_counts().head(60))

# =========================
# REPORT
# =========================
report_rows = []

for rule_name, condition, func in punctuation_rules:
    report_rows.append({
        "Rule_Name": rule_name,
        "Candidate_Freq": len(rule_candidates.get(rule_name, [])),
        "Injected_Count": rule_usage.get(rule_name, 0)
    })

report_df = pd.DataFrame(report_rows).sort_values("Injected_Count", ascending=False)

summary_df = pd.DataFrame({
    "Metric": [
        "Dataset Size",
        "Target Punctuation Errors",
        "Injected Punctuation Errors",
        "Total Rules",
        "Rules With Matches",
        "Rules Used"
    ],
    "Value": [
        len(df),
        target_count,
        modified,
        len(punctuation_rules),
        len(rule_candidates),
        (report_df["Injected_Count"] > 0).sum()
    ]
})

with pd.ExcelWriter(REPORT_PATH) as writer:
    summary_df.to_excel(writer, sheet_name="Summary", index=False)
    report_df.to_excel(writer, sheet_name="Rule_Usage", index=False)

print("Report saved to:", REPORT_PATH)

print("\nTop used rules:")
print(report_df.head(30).to_string(index=False))
```

```python
import pandas as pd

INPUT_PATH = "/kaggle/working/51k_after_punctuation_injection.csv"

df = pd.read_csv(INPUT_PATH)
df["sentence"] = df["sentence"].astype(str)
df["error_label"] = df["error_label"].astype(str)

# labels containing punctuation error
punct_df = df[
    df["error_label"].apply(
        lambda x: "11" in [i.strip() for i in str(x).split(",")]
    )
].copy()

print("Total punctuation error sentences:", len(punct_df))

print("\nLabel combinations containing 11:")
print(punct_df["error_label"].value_counts().sort_index())

# =========================
# PRINT EXAMPLES
# =========================
def show_examples(label_combo, n=5):
    subset = df[df["error_label"] == label_combo]

    print("\n" + "=" * 100)
    print(f"LABEL {label_combo}")
    print("=" * 100)

    if len(subset) == 0:
        print("No examples found.")
        return

    sample = subset.sample(
        n=min(n, len(subset)),
        random_state=42
    )

    for i, row in enumerate(sample.itertuples(), 1):
        print(f"\nNo. {i}")
        print("Sentence :", row.sentence)
        print("Label    :", row.error_label)

# only punctuation
show_examples("11", n=10)

# punctuation + single old error
for combo in ["2,11", "3,11", "4,11", "5,11", "6,11", "9,11"]:
    show_examples(combo, n=5)

# all existing combos containing punctuation
all_punct_combos = sorted(
    punct_df["error_label"].unique(),
    key=lambda x: [int(i) for i in x.split(",")]
)

print("\n" + "=" * 100)
print("ALL LABEL COMBINATIONS CONTAINING 11")
print("=" * 100)

for combo in all_punct_combos:
    show_examples(combo, n=3)
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_id = "google/gemma-3-1b-it"

tokenizer = AutoTokenizer.from_pretrained(model_id)

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.bfloat16,
    device_map="auto"
)

print("Model loaded successfully!")
```

```python
messages = [
    {"role": "user", "content": "Who are you?"}
]

inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
).to(model.device)

outputs = model.generate(
    **inputs,
    max_new_tokens=50
)

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
